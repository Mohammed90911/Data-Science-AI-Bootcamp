## Olympic Sports and Medals, 1896-2014

### Combining The Data

```python
countries_df = pd.read_csv('dictionary.csv')
summer_df = pd.read_csv('summer.csv')
winter_df = pd.read_csv('winter.csv')

summer_df = summer_df.rename(columns={'Country': 'Code'})
winter_df = winter_df.rename(columns={'Country': 'Code'})

summer_countries_df = pd.merge(countries_df, summer_df, on='Code', how='inner')
summer_countries_df['Season'] = 'Summer'

winter_countries_df = pd.merge(countries_df, winter_df, on='Code', how='inner')
winter_countries_df['Season'] = 'Winter'

all_df = pd.concat([summer_countries_df, winter_countries_df], axis='index')
all_df.head(3)
```

```python
from nbresult import ChallengeResult

result = ChallengeResult(
    'all_df', all_df_shape=all_df.shape, all_df_columns=set(all_df.columns)
)
result.write()
print(result.check())
```

### Top Countries Analysis

```python
since_1984 = all_df[all_df['Year'] >= 1984]

top_10_df = (
    since_1984.groupby('Country')[['Medal']]
    .count()
    .rename(columns={'Medal': 'Medal Count'})
    .sort_values(by='Medal Count', ascending=False)
    .head(10)
)

top_10_df
```

```python
from nbresult import ChallengeResult

result = ChallengeResult(
    'olympic_games',
    top_country_1=top_10_df.iloc[0]['Medal Count'],
    top_country_10=top_10_df.iloc[9]['Medal Count'],
)
result.write()
print(result.check())
```

```python

season_df = (
    since_1984.groupby(['Country', 'Season'])[['Medal']].count().unstack()
)

season_df.columns = season_df.columns.droplevel(0)
season_df = season_df.fillna(0)

season_df['Total'] = season_df['Summer'] + season_df['Winter']
top_10_season_df = (
    season_df.sort_values(by='Total', ascending=False)
    .head(10)[['Summer', 'Winter', 'Total']]
)

top_10_season_df
```

```python
from nbresult import ChallengeResult

result = ChallengeResult(
    'olympic_games_season',
    top_country_season_shape=top_10_season_df.shape,
    top_country_1_summer=top_10_season_df.iloc[0]['Summer'],
    top_country_10_winter=top_10_season_df.iloc[9]['Winter'],
)
result.write()
print(result.check())
```

```python
events_df = since_1984.drop_duplicates(
    subset=['Year', 'Season', 'City', 'Sport', 'Discipline', 'Event', 'Country', 'Medal']
)

top_10_events_df = (
    events_df.groupby('Country')[['Event']]
    .count()
    .rename(columns={'Event': 'Event Count'})
    .sort_values(by='Event Count', ascending=False)
    .head(10)
)

top_10_events_df
```

```python
from nbresult import ChallengeResult

result = ChallengeResult(
    'olympic_games_event',
    top_country_event_shape=top_10_events_df.shape,
    top_country_1_event=top_10_events_df.iloc[0]['Event Count'],
    top_country_10_event=top_10_events_df.iloc[9]['Event Count'],
)
result.write()
print(result.check())
```

```python
top_10_combined = (
    pd.concat([top_10_df, top_10_events_df], axis=1)
    .dropna(subset=['Medal Count', 'Event Count'], how='all')
    .sort_values(by='Medal Count', ascending=False)
)

top_10_combined
```

```python
from nbresult import ChallengeResult

result = ChallengeResult(
    'olympic_games_combined',
    top_combined_shape=top_10_combined.shape,
    top_combined_1_event=top_10_combined.iloc[0]['Event Count'],
    top_combined_1_medal=top_10_combined.iloc[0]['Medal Count'],
    top_combined_10_event=top_10_combined.iloc[9]['Event Count'],
    top_combined_10_medal=top_10_combined.iloc[9]['Medal Count'],
)
result.write()
print(result.check())
```
